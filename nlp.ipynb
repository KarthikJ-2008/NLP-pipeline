```python
text = "Natural Language Processing is a subfield of artificial intelligence focused on processing human language—both text and speech—so machines can interact with people naturally. It combines rule‑based linguistic models with statistical and neural approaches to analyze meaning, structure, and context in language data. NLP is used in applications such as machine translation, sentiment analysis, speech recognition, and chatbots."
```

```python
# Tokenization

from nltk import word_tokenize 

words = word_tokenize(text)
print(words)
print("Total number of words in the text:", len(words))
```

```python
# Normalization

elist=[]
for i in words:
    if i.isalpha():
        elist.append(i.lower())
print(elist)
print("Total number of words after normalization:", len(elist))
```

```python
# Stopwords removal

from nltk.corpus import stopwords
sw=stopwords.words('english')
print(sw)
```

```python
elist1=[]
for i in elist:
    if i not in sw:
        elist1.append(i)
print(elist1)
print("Total number of words after stopword removal:", len(elist1))
```

```python
# Parts of Speech Tagging

from nltk import pos_tag

tag=pos_tag(elist1)
print(tag)
```

```python
# Named Entity Recognition

from nltk import ne_chunk

tree = ne_chunk(tag, binary=True)

for i in tree:
    print(i, end=" ")
# tree.draw()
```

```python
# Lemmatization

from nltk import WordNetLemmatizer
lemma = WordNetLemmatizer()
elist2=[]
for i in elist1:
    elist2.append(lemma.lemmatize(i))
print(elist2)
```
